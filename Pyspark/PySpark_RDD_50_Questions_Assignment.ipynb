### Q1. Multiply every number by 2

Dataset:
[5, 10, 15, 20, 25, 30]

Problem:
Create an RDD from the dataset and multiply every number by 2.

Problem:

image.png

We are treating each tuple conceptually as a key-value pair, but it is NOT a Python dictionary.

```python
sentences = [
    "spark is fast",
    "spark is powerful",
    "spark is easy",
    "hadoop is distributed",
    "spark is fast"
]
sentence_rdd = sc.parallelize(sentences)
words = sentence_rdd.flatMap(lambda x: x.split())
word_counts = words.map(lambda x: (x, 1)).reduceByKey(lambda a, b: a + b)
print(*word_counts.collect(), sep="\n")
```

### Q50. Complete business analysis

Dataset:
orders = [
    ("O1", "Laptop", 2, 50000),
    ("O2", "Mobile", 5, 20000),
    ("O3", "Laptop", 1, 50000),
    ("O4", "Tablet", 3, 15000),
    ("O5", "Mobile", 2, 20000),
    ("O6", "Laptop", 4, 50000)
]
order_rdd = sc.parallelize(orders)

Problem:
Perform the following complete analysis:
1. Calculate revenue for every order.
2. Keep only orders with revenue greater than 50000.
3. Calculate total revenue for each product.
4. Find the product with the highest revenue.
5. Save the final result into a single output partition.

```python
orders = [
    ("O1", "Laptop", 2, 50000),
    ("O2", "Mobile", 5, 20000),
    ("O3", "Laptop", 1, 50000),
    ("O4", "Tablet", 3, 15000),
    ("O5", "Mobile", 2, 20000),
    ("O6", "Laptop", 4, 50000)
]
order_rdd = sc.parallelize(orders)
revenue_rdd = order_rdd.map(lambda x: (x[0], x[1], x[2] * x[3]))

retained = revenue_rdd.filter(lambda x: x[2] > 50000)
print("Orders retained:")
print(*retained.collect(), sep="\n")

product_revenue = (
    retained
    .map(lambda x: (x[1], x[2]))
    .reduceByKey(lambda a, b: a + b)
)

print("Product Revenue:")
print(*product_revenue.collect(), sep="\n")

highest = product_revenue.max(key=lambda x: x[1])
print("Highest Revenue Product:")
print(highest)

final_rdd = sc.parallelize([highest]).coalesce(1)
final_rdd.saveAsTextFile("output/final_result")

print("Output partitions:", final_rdd.getNumPartitions())
```
