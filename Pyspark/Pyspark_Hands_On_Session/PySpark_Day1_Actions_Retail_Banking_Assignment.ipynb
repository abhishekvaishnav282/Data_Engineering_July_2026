## 1. Start Spark and create the retail DataFrame

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.types import *

spark = SparkSession.builder.appName("RetailPySparkTraining").getOrCreate()

retail_data = [
    (1001,"Laptop","Electronics",65000.0,2,0.10,"Bengaluru","2026-09-01"),
    (1002,"Phone","Electronics",30000.0,3,0.05,"Hyderabad","2026-09-02"),
    (1003,"Shoes","Fashion",4500.0,5,0.20,"Chennai","2026-09-02"),
    (1004,"Watch","Fashion",8000.0,2,0.15,"Bengaluru","2026-09-03"),
    (1005,"TV","Electronics",55000.0,1,0.10,"Pune","2026-09-03"),
    (1006,"Chair","Furniture",7000.0,4,0.05,"Hyderabad","2026-09-04"),
    (1007,"Table","Furniture",12000.0,2,0.10,"Chennai","2026-09-04"),
    (1008,"Headphones","Electronics",3500.0,6,0.00,"Bengaluru","2026-09-05"),
    (1009,"Bag","Fashion",2500.0,8,0.10,"Pune","2026-09-05"),
    (1010,"Sofa","Furniture",40000.0,1,0.12,"Bengaluru","2026-09-06")
]
cols=["order_id","product","category","price","quantity","discount","city","order_date"]
df=spark.createDataFrame(retail_data,cols)
df.show(truncate=False)
```

## 2. Inspect the DataFrame
`show`, `printSchema`, `columns`, and `dtypes` help us understand incoming retail data.

```python
df.show(5, truncate=False)
df.printSchema()
print(df.columns)
print(df.dtypes)
```

## 3. Actions in PySpark
Actions trigger execution and return a result to the driver or write output.

### `show()` — display rows

```python
df.show(5, truncate=False)
```

### `count()` — count retail orders

```python
print("Total orders:", df.count())
```

### `first()` and `head()` — retrieve initial rows

```python
print("First order:", df.first())
print("First 3 orders:", df.head(3))
```

### `take()` — retrieve N rows

```python
for row in df.take(3):
    print(row)
```

### `collect()` — bring all rows to the driver
Use only for small results.

```python
rows=df.collect()
print(rows)
```

### `foreach()` — execute logic for every row

```python
df.select("order_id","product").foreach(lambda row: print(row.order_id, row.product))
```

### `reduce()` on an RDD — calculate total quantity

```python
total_qty=df.select("quantity").rdd.map(lambda r:r[0]).reduce(lambda a,b:a+b)
print("Total quantity:",total_qty)
```

### `countByValue()` — frequency of retail categories

```python
print(df.select("category").rdd.map(lambda r:r[0]).countByValue())
```

### `isEmpty()` — check whether data exists

```python
print("Is DataFrame empty?", df.isEmpty())
```

### `toLocalIterator()` — iterate without collecting everything at once

```python
for row in df.select("order_id","product").toLocalIterator():
    print(row)
```

### Challenge: Which actions are dangerous on a very large banking dataset?

| Action | Why it is risky | Safer alternative |
|---|---|---|
| `collect()` | Pulls **every row** from all executors into driver memory. Millions of transactions → driver `OutOfMemoryError` and crash. | `show(n)`, `take(n)`, `limit(n).collect()`, or write the result to storage |
| `toPandas()` | Same as `collect()`, plus conversion to a pandas DataFrame on a single machine. | Aggregate or filter first, then convert only the small result |
| `countByValue()` / `countByKey()` | Returns a Python dict to the driver. A high-cardinality column such as `customer_id` or `transaction_id` gives millions of keys → driver memory pressure. | `groupBy("col").count()`, which stays distributed |
| `show(df.count())` / `head(large_n)` / `take(large_n)` | A large `n` again ships huge data to the driver. `show(df.count())` also scans the whole dataset twice. | Keep `n` small for inspection |
| `count()` | Safe for memory, but it runs a **full scan** of the dataset. Calling it repeatedly on uncached data is slow and costly. | Cache if reused, or use `isEmpty()` / `limit(1)` when you only need to know whether data exists |
| `reduce()` | Result is small, but a plain Python lambda on an RDD is slower than built-in aggregates and can overflow or lose precision on money. | `bank_df.agg(F.sum("amount"))` (use `DecimalType` for currency) |
| `foreach()` | Runs on executors. Side effects like `print` or calling external systems (e.g. a DB insert per transaction) can overload them or run twice on task retry, which is a problem for financial data. | `foreachPartition()` with batched, idempotent writes |
| `toLocalIterator()` | Safer than `collect()` because it fetches one partition at a time, but a very large partition can still exhaust driver memory, and it is slow and sequential. | Process in a distributed way, or write to storage |
| `write` with `mode("overwrite")` | Can silently **replace existing production transaction data**. | Use `append` or partitioned writes, and write to a staging path first |

**Key idea:** actions that return data to the **driver** (`collect`, `toPandas`, `countByValue`, large `take`/`head`) are the most dangerous. Banking data is huge and sensitive, so filter or aggregate on the cluster first and bring back only small results.
