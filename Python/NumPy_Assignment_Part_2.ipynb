# Part 1: Dataset Creation

### Dataset Setup

Create the `stores`, `categories`, `sales_data` and `cost_data` arrays.
Rows represent stores, columns represent product categories.

```python
import numpy as np

stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
```

Classify every sales value as `"High"` (at least 5,00,000), `"Medium"` (2,50,000 to

# Part 9: Target-Based Analysis

## Question 48: Target Comparison

Compare every store's category sales against `sales_targets` using NumPy broadcasting.

`sales_targets` has shape `(6,)` and `sales_data` has shape `(5, 6)`. Broadcasting
stretches the single target row down across all five stores at zero memory cost.

The important question is:

Does (6,) match the rows or the columns?

Answer: it matches the last dimension, which here is the 6 columns.
NumPy aligns shapes from the RIGHT side:

You might think:

"There are 5 stores, so shouldn't (5,) match the 5 rows?"

No. This is the important rule.

NumPy broadcasting does not automatically decide:

"Oh, 5 means rows."

Instead, broadcasting compares dimensions from the rightmost side.

The easiest rule to remember

When broadcasting:

NumPy matches dimensions from RIGHT → LEFT.

```python
sales_targets = np.array([
    800000,
    450000,
    320000,
    300000,
    200000,
    230000
])

print("sales_data shape   :", sales_data.shape)
print("sales_targets shape:", sales_targets.shape)

comparison = sales_data - sales_targets

print("\nSales minus target for every store:")
print(comparison)

print()
print(f"{'Store':<12}" + "".join(f"{c:>13}" for c in categories))
for store, row in zip(stores, comparison):
    print(f"{store:<12}" + "".join(f"{value:>13}" for value in row))
```

**The difference:** `flatten()` always returns a **copy**, so editing it never touches
the original. `ravel()` returns a **view** whenever possible, so editing it can change
the original array. `ravel()` is faster and uses less memory; `flatten()` is safer.

## Question 93: Replace with Company Average

Replace missing values with the overall average of the available sales values using

```python
company_average = np.nanmean(sales_with_missing)

sales_average_filled = sales_with_missing.copy()
print(np.isnan(sales_average_filled))
sales_average_filled[np.isnan(sales_average_filled)] = company_average

print(f"Company average of available values: {company_average:,.2f}")

print("\nAfter replacing missing values with the company average:")
print(np.round(sales_average_filled, 2))

print("\nAny missing values remaining?", np.isnan(sales_average_filled).any())
```

## Question 94: Replace with Category Average

Replace each missing value with the average sales of its own product category.

`np.nanmean(axis=0)` gives one average per column. `np.take` then picks the correct
category average for each missing position using its column index.

For example:

np.where(missing_positions)

could give:

rows    = [0, 1, 2, 3]
columns = [1, 2, 0, 3]

The expression:

np.where(missing_positions)[1]

takes the second part, which is the column indexes:

[1, 2, 0, 3]
np.take(category_averages, [1, 2, 0, 3])

```python
category_averages = np.nanmean(sales_with_missing, axis=0)

print("Category averages (ignoring missing values):")
for category, average in zip(categories, category_averages):
    print(f"  {category:<12}: {average:,.2f}")

sales_category_filled = sales_with_missing.copy()
missing_positions = np.isnan(sales_category_filled)
print(missing_positions)
print(sales_category_filled[missing_positions] )
sales_category_filled[missing_positions] = np.take(category_averages, np.where(missing_positions)[1])

print("\nAfter replacing missing values with category averages:")
print(np.round(sales_category_filled, 2))

print("\nAny missing values remaining?", np.isnan(sales_category_filled).any())
```
