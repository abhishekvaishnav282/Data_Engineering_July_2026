# Part 1: Dataset Creation

### Dataset Setup

Create the `stores`, `categories`, `sales_data` and `cost_data` arrays.
Rows represent stores, columns represent product categories.

```python
import numpy as np

stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
```

Display the number of dimensions, shape, total number of values, data type and memory

## Question 3: Sales Data Type Conversion

Convert `sales_data` into `float32` and `float64`, then compare the memory used by both.

`float32` uses 4 bytes per value and `float64` uses 8, so the same data occupies twice
the memory in `float64`.

```python
sales_float32 = sales_data.astype(np.float32)
sales_float64 = sales_data.astype(np.float64)

print("Original dtype :", sales_data.dtype, "|", sales_data.nbytes, "bytes")
print("float32 dtype  :", sales_float32.dtype, "|", sales_float32.nbytes, "bytes")
print("float64 dtype  :", sales_float64.dtype, "|", sales_float64.nbytes, "bytes")

print("\nfloat64 uses", sales_float64.nbytes - sales_float32.nbytes, "bytes more than float32")

print("\nfloat32 array:")
print(sales_float32)
```

## Question 33: Maximum Individual Profit

Find the highest profit from any single store-category combination and display the
profit amount, store name and category. Use `np.unravel_index()`.

flat number back into a row and column pair.

```python
print("profit_data :",profit_data)
flat_position = np.argmax(profit_data)
print("flat_position :",flat_position)
row, column = np.unravel_index(flat_position, profit_data.shape)

print("Flat position   :", flat_position)
print("Row, column     :", (row, column))
print("Profit amount   :", profit_data[row, column])
print("Store name      :", stores[row])
print("Product category:", categories[column])
```

## Question 47: Category Performance Classification

Classify every sales value as `"High"` (at least 5,00,000), `"Medium"` (2,50,000 to
4,99,999) or `"Low"` (below 2,50,000). Use `np.select()`.

`np.select()` checks the condition list in order and applies the first match, so the
conditions do not need to exclude one another manually.

```python
conditions = [
    sales_data >= 500000,
    sales_data >= 250000,
    sales_data < 250000
]

labels = ["High", "Medium", "Low"]

print("conditions :",conditions)
classification = np.select(conditions, labels, default="Unknown")

print("Classification matrix:")
print(classification)

print()
print(f"{'Store':<12}" + "".join(f"{c:>13}" for c in categories))
for store, row in zip(stores, classification):
    print(f"{store:<12}" + "".join(f"{value:>13}" for value in row))

print("np.select() does NOT process one entire Boolean array and then move to the next array. Instead, conceptually, it looks at the same position across all condition arrays, decides which condition is the first True, and puts the corresponding label at that position.")
values, counts = np.unique(classification, return_counts=True)
for label, count in zip(values, counts):
    print(f"{label:<8}: {count}")
```
