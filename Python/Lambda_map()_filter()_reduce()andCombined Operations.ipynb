Using `reduce()`, calculate the total payroll directly from the dictionaries.reduce() walks through the list one number at a time, carrying an accumulator (a) forward. At each step, it checks if the current number (b) is even — if yes, adds it to the accumulator; if no, keeps the accumulator unchanged.
