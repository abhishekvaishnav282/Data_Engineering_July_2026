Create four separate functions: `add()`, `subtract()`, `multiply()`, `divide()`.

**Requirements**
- Select an operation and call the appropriate function.
- Handle division by zero.
- Print the calculated result.

````python

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: division by zero is not allowed"
    return a / b

def calculate(a, b, operation):
    if operation == "add":
        return add(a, b)
    elif operation == "subtract":
        return subtract(a, b)
    elif operation == "multiply":
        return multiply(a, b)
    elif operation == "divide":
        return divide(a, b)
    else:
        return "Error: unknown operation"

print("Result:", calculate(12, 4, "divide"))

print("Result:", calculate(9, 0, "divide"))


print("Result:", calculate(7, 3, "multiply"))
````

- Return the factorial and handle negative input.

````python


def calculate_factorial(number):

    if number < 0:
        return "Error: factorial is not defined for negative numbers"

    result = 1

    for value in range(1, number + 1):
        result = result * value
    return result

# Test 1
print("Factorial:", calculate_factorial(5))

# Test 2 - 0! is 1 by definition
print("Factorial:", calculate_factorial(0))

# Test 3 - negative input is rejected
print("Factorial:", calculate_factorial(-4))
````

### Assignment 32: Function Returning a Dictionary

Write a function named `analyze_text()`.

**Requirements**
- Accept a string and return a dictionary containing total characters,
  words, vowels, digits, and spaces.
- Display every result clearly.

````python


def analyze_text(text):
    vowels = "aeiouAEIOU"

    vowel_count = 0
    digit_count = 0
    space_count = 0

    # A single pass over the text collects every count
    for character in text:
        if character in vowels:
            vowel_count += 1
        if character.isdigit():
            digit_count += 1
        if character == " ":
            space_count += 1

    # Returning a dictionary keeps the results clearly labelled
    return {
        "total_characters": len(text),
        "total_words": len(text.split()),
        "total_vowels": vowel_count,
        "total_digits": digit_count,
        "total_spaces": space_count
    }

# Test 1
result = analyze_text("Python 3 Functions are powerful in 2026")
print(result)

print("-" * 35)

# Test 2
result = analyze_text("Edukron Academy 2024")
for key, value in result.items():
    print(f"{key.replace('_', ' ').title()} : {value}")
````

### Assignment 34: Prime Number Function

Write a function named `is_prime()`.

**Requirements**
- Accept one integer and return `True` when the number is prime.
- Use the function to print all prime numbers from 1 to 100.

````python


def is_prime(number):

    if number < 2:
        return False

    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True


for value in [1, 2, 17, 20, 97]:
    print(f"{value} -> prime: {is_prime(value)}")

print()
primes = []
for number in range(1, 101):
    if is_prime(number):
        primes.append(number)

print("Prime numbers from 1 to 100:")
print(primes)
````

### Assignment 35: Fibonacci Function

Write a function named `generate_fibonacci()`.

**Requirements**
- Accept the number of terms and return the series as a list.
- Handle zero and negative term counts.
- Do not use recursion.

````python

def generate_fibonacci(terms):

    if terms <= 0:
        return []

    series = [0]
    if terms == 1:
        return series

    series.append(1)

    while len(series) < terms:
        next_term = series[-1] + series[-2]
        series.append(next_term)
    return series


print("7 terms :", generate_fibonacci(7))


print("12 terms:", generate_fibonacci(12))


print("0 terms :", generate_fibonacci(0))
print("-5 terms:", generate_fibonacci(-5))
````

````python
# Assignment 67: Filter Prime Numbers

def is_prime(number):
    # Anything below 2 (including negatives and 0) is not prime
    if number < 2:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True

numbers = [-5, 0, 1, 2, 3, 4, 9, 11, 15, 17, 20, 23]
primes = list(filter(is_prime, numbers))

print("Original numbers:", numbers)
print("Prime numbers   :", primes)

# Test 2
print("Prime numbers   :", list(filter(is_prime, [29, 30, 31, 33])))
````

### Assignment 92: Closure-Based Bank Balance

**Requirements**
- Store the balance in the outer function.
- Define inner functions for deposit, withdrawal, and balance enquiry.
- Use `nonlocal` to update the balance and reject over-limit withdrawals.

````python


def create_bank_account(opening_balance=0):
    balance = opening_balance   # private to this closure

    def deposit(amount):
        nonlocal balance
        if amount <= 0:
            return "Error: deposit must be greater than zero"
        balance += amount
        return f"Deposited {amount}. New balance: {balance}"

    def withdraw(amount):
        nonlocal balance
        if amount <= 0:
            return "Error: withdrawal must be greater than zero"

        if amount > balance:
            return f"Error: insufficient funds. Available balance: {balance}"
        balance -= amount
        return f"Withdrew {amount}. New balance: {balance}"

    def check_balance():
        return f"Current balance: {balance}"

    # Returning all three functions gives controlled access to 'balance'
    return deposit, withdraw, check_balance

deposit_0, withdraw_1, check_balance_2 = create_bank_account(1000)
deposit, withdraw, check_balance = create_bank_account(1000)

print(check_balance_2())
print(deposit_0(500))
print(withdraw_1(300))
print(withdraw_1(5000))      # rejected
print(deposit_0(-50))        # rejected
print(check_balance_2())

print()

print(check_balance())
print(deposit(500))
print(withdraw(300))
print(withdraw(5000))        # rejected
print(deposit(-50))          # rejected
print(check_balance())
````

Create a decorator named `execution_logger()`.

````python


def execution_logger(original_function):
    def wrapper(*args, **kwargs):
        print("Starting function...")

        # Capture the return value so it is not lost
        result = original_function(*args, **kwargs)
        print(f"Result: {result}")
        print("Function completed.")
        return result

    return wrapper

@execution_logger
def add_numbers(a, b):
    return a + b

# Test 1 - the decorator runs around the function
returned_value = add_numbers(10, 20)
print("Value received by the caller:", returned_value)

print()

# Test 2
add_numbers(7.5, 2.5)
````

### Assignment 99: LEGB Name Resolution

**LEGB Order:** `L -> Local`, `E -> Enclosing`, `G -> Global`, `B -> Built-in`

````python

value = "GLOBAL value"

# ---------- Case 1: all three scopes present ----------
def outer():
    value = "ENCLOSING value"   # E - Enclosing scope

    def inner():
        value = "LOCAL value"   # L - Local scope
        print("  inner() sees :", value)

    inner()
    print("  outer() sees :", value)

print("Case 1 - local, enclosing and global all defined")
print("  global sees  :", value)
outer()

# ---------- Case 2: the LOCAL declaration is removed ----------
def outer_no_local():
    value = "ENCLOSING value"

    def inner():
        # No local 'value', so Python moves outward to the enclosing scope
        print("  inner() sees :", value)

    inner()

print("\nCase 2 - local removed, so the ENCLOSING value is used")
outer_no_local()

# ---------- Case 3: local AND enclosing removed ----------
def outer_no_enclosing():
    def inner():
        # No local, no enclosing, so Python falls back to the global scope
        print("  inner() sees :", value)

    inner()

print("\nCase 3 - local and enclosing removed, so the GLOBAL value is used")
outer_no_enclosing()

# ---------- Case 4: built-in scope ----------
print("\nCase 4 - built-in scope")
print("  len is found in the BUILT-IN scope:", len("Edukron"))

# EXPLANATION - LEGB lookup order:
# L - Local     : names defined inside the current function
# E - Enclosing : names in any outer function that wraps this one
# G - Global    : names defined at module level
# B - Built-in  : names Python provides automatically (len, print, sum, ...)
# Python stops at the FIRST scope where the name is found.
````
