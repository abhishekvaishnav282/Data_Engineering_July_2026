In one line: @property is useful when you want method logic but attribute-style access.
So why do we need @property?

@property is mainly for attribute-style access without arguments; normal methods are for operations that may need arguments.

```python
class CreateAAccountClass:
    bank_name = 'City Bank'
    def __init__(self, account_ref='ACC-1001', balance=1000):
        self.account_ref = account_ref
        self._balance = float(balance)
    @property
    def balance(self):
        return self._balance
    def deposit(self, amount):
        self._balance += amount
        return self._balance
    def __str__(self):
        return f'{self.account_ref}: balance={self._balance:.2f}'

acc = CreateAAccountClass()
acc2 = CreateAAccountClass('ACC-1002',2000)
print(acc, acc.deposit(500))
print(acc.balance)
print(acc2)
print(acc2.deposit(15242))
print(acc2.balance)
print(acc2)
```

@property controls reading; @amount.setter controls changing.

```python
class Withdrawal:
    def __init__(self, amount):
        self._amount = amount

    @property
    def amount(self):
        return self._amount
    
    @amount.setter
    def amount(self, value):
        if value < 0:
            raise ValueError('amount cannot be negative')
        self._amount = value

w = Withdrawal(500)
print(w.amount)
w.amount = 700
print(w.amount)
```

This example demonstrates @classmethod, which is useful when you want to create an object using an alternative way of providing data.
self → current OBJECT
cls  → current CLASS
cls  → Account
data → "ACC-1012,1500"

return cls(ref, float(balance)) ===> return Account(ref, float(balance))

Account('ACC-1012', 1500.0)
          ↓
       __init__()
          ↓
account_ref = "ACC-1012"
balance = 1500.0

```python
class Account:
    def __init__(self, account_ref, balance):
        self.account_ref = account_ref
        self.balance = balance
    @classmethod
    def from_string(cls, data):
        ref, balance = data.split(',')
        return cls(ref, float(balance))

acc = Account.from_string('ACC-1012,1500')
print(acc.account_ref, acc.balance)
```

```python
class PricingRule:
    def apply(self, amount):
        return amount

class PercentageFeeRule(PricingRule):
    def __init__(self, percent):
        self.percent = percent
    def apply(self, amount):
        return amount + amount * self.percent / 100

class PricingEngine:
    def __init__(self, rules):
        self.rules = rules
    def price(self, amount):
        for rule in self.rules:
            amount = rule.apply(amount)
        return amount

engine = PricingEngine([PercentageFeeRule(2)])
print(engine.price(1000))
```
